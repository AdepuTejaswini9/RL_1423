**Problem-1**


```python
# Assignment 1: Calculate Undiscounted and Discounted Returns

# Accept number of time steps
n = int(input("Enter the number of time steps/episodes: "))

# Accept reward values
rewards = []

for i in range(n):
    reward = float(input(f"Enter reward for time step {i + 1}: "))
    rewards.append(reward)

# Accept discount factor
gamma = float(input("Enter the discount factor gamma (γ): "))

# Calculate undiscounted return
undiscounted_return = sum(rewards)

# Calculate discounted return
discounted_return = 0

for i in range(n):
    discounted_return += (gamma ** i) * rewards[i]

# Display results
print("\n----- Results -----")
print("Entered reward values:", rewards)
print("Undiscounted return:", undiscounted_return)
print("Discounted return:", round(discounted_return, 4))
```

**Problem-2:**

```python
from grid_env_det import GridWorldMDP

# Create MDP environment
env = GridWorldMDP()

print("===== 2 x 3 GRID WORLD MDP =====\n")

# a. State Space
print("State Space:")
print(env.states)

# b. Action Space
print("\nAction Space:")
print(env.actions)

# c. Reward Function
print("\nReward Function:")
print("Normal State Reward =", env.normal_reward)
print("Goal Reward =", env.goal_reward)
print("Obstacle Reward =", env.obstacle_reward)

# d. Discount Factor
print("\nDiscount Factor (gamma) =", env.gamma)

# e. Transition Function
print("\nTransition Function:")

for state in env.states:
    if state == "G":
        continue

    for action in env.actions:
        next_state = env.transition(state, action)
        print(f"{state} + {action} -> {next_state}")

# f. Estimate next state
print("\n===== NEXT STATE ESTIMATION =====")

current_state = input("Enter current state (S0, S1, S3, S4): ")
action = input("Enter action (up, down, left, right): ").lower()

try:
    next_state = env.transition(current_state, action)
    reward = env.reward(current_state, action, next_state)

    print("\nCurrent State :", current_state)
    print("Action        :", action)
    print("Next State    :", next_state)
    print("Reward        :", reward)

    if env.is_terminal(next_state):
        print("Status        : Goal state reached")

except ValueError as e:
    print("Error:", e)
```
